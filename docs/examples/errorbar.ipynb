# Errorbar

The interactive errorbar function is available as {func}`mpl_interactions.pyplot.interactive_errorbar` or as `mpl_interactions.ipyplot.errorbar`. Any of `x`, `y`, `yerr`, and `xerr` can be functions of the slider parameters, making it easy to explore how error bars change.

```python
%matplotlib ipympl
import matplotlib.pyplot as plt
import numpy as np

import mpl_interactions.ipyplot as iplt
```

## Basic example

Pass a function for `y` and `yerr` with slider parameters to interactively control the plot.

```python
x = np.linspace(0, 2 * np.pi, 50)


def y_func(x, tau):
    return np.sin(x * tau)


def yerr_func(tau):
    return np.full(50, 0.05 * tau)


fig, ax = plt.subplots()
controls = iplt.errorbar(x, y_func, yerr=yerr_func, tau=(1, 10, 100))
```

## Functions for x, y, xerr, and yerr

All four data arguments can be functions. The function for `y` receives the evaluated `x` array as its first argument, while `x`, `xerr`, and `yerr` receive only the slider parameters.

```python
def x_func(tau):
    return np.linspace(0, 2 * np.pi * tau, 50)


def y_func(x, tau):
    return np.sin(x)


def yerr_func(tau):
    return np.full(50, 0.1 * tau)


def xerr_func(tau):
    return np.full(50, 0.05 * tau)


fig, ax = plt.subplots()
controls = iplt.errorbar(
    x_func, y_func, yerr=yerr_func, xerr=xerr_func, tau=(1, 5, 100), xlim="auto"
)
```

## Styling

Errorbar-specific kwargs (`capsize`, `ecolor`, `elinewidth`, etc.) and Line2D kwargs (`color`, `linewidth`, `markersize`, etc.) are passed through to `ax.errorbar`.

```python
x = np.linspace(0, 2 * np.pi, 30)


def y_func(x, tau):
    return np.cos(x * tau)


def yerr_func(tau):
    return np.full(30, 0.1 * tau)


fig, ax = plt.subplots()
controls = iplt.errorbar(
    x,
    y_func,
    yerr=yerr_func,
    tau=(1, 10, 100),
    fmt="o",
    markersize=8,
    capsize=4,
    ecolor="gray",
    color="C0",
)
```
